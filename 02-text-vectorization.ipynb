# Different Techniques to Vectorize Text Data

## Bag of Words (BoW)

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer()
```

```python
df = pd.DataFrame({
    "Text":['Python is a programming language Python',
            'Python can be used for data science',
            'Data science is an interdisciplinary field',
            'Machine learning is a subset of data science',
            'Python has many libraries for data science'],
    "Label":[1,1,0,0,1]
})
```

```python
df
```

```python
bog = cv.fit_transform(df['Text'])
```

```python
bog
```

```python
cv.vocabulary_
```

```python
bog[0].toarray()
```

```python
df['Text'][0]
```

## N-Grams

```python
cv = CountVectorizer(ngram_range=(2,2))
bog = cv.fit_transform(df['Text'])
```

```python
cv.vocabulary_
```

```python
bog[0].toarray()
```

```python
bog.toarray()
```

```python
len(bog.toarray())
```

## TF-IDF (Term Frequency-Inverse Document Frequency)

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_bog = tfidf_vectorizer.fit_transform(df['Text']).toarray()
```
